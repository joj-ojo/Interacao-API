# 🐱 Versão 1: uma imagem de gato

Faz uma requisição à API [The Cat API](https://thecatapi.com/), lê o endereço da imagem na resposta JSON, baixa e salva em `imagens/cat/cat.jpg`.

> No Colab os arquivos ficam em `/content` e **são apagados quando a sessão termina**. Baixe o que quiser guardar pelo painel de arquivos (ícone de pasta, à esquerda).

## 1. Importações

```python
import os # para interagir com osistema operacional
import requests # para fazer requisições htp e interagir com a API
from IPython.display import Image, display #para visualizar as imagens
```

## 2. Pasta de saída

```python
pasta = os.path.join(os.getcwd(), 'imagens/cat') #armazena na variavel pasta o caminhoda pasta imagens/cat que o comando de pasta padrão criou "getcwd = get current working directory (diretorio onde estamos)"
os.makedirs(pasta, exist_ok=True)
caminho = os.path.join(pasta, 'cat.jpg') #define o nome da imagem salva e onde será salva
```

## 3. Requisição à API e download da imagem

```python
url = 'https://api.thecatapi.com/v1/images/search' #endereço de conexão com a api, entra na API e busca o endereço dela
response = requests.get(url) #armazena a resposta da requisição que foi buscada na url
data = response.json() #converte a resposta para json
img_url = data[0]['url'] #neste caso como a api está organizada em lista, precisa indicar o indice da lista
img_data = requests.get(img_url).content #faz um novo get para buscar a imagem
with open(caminho, 'wb') as handler: #converte o conteudo para imagem
    handler.write(img_data) #função que faz a conversão e guarda na pasta
print(f"Imagem do gato salva em {caminho}!") # imprime
```

## 4. Visualizar a imagem

```python
display(Image(filename=caminho, width=400)) #imprime a imagem no caminho e tamanho especificado
```
